## Imports

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Solve For Forces in Each Member

```python

def truss(theta_A, theta_C, P):

    # change the angles to radians
    A = np.radians(theta_A)
    C = np.radians(theta_C)

    # make the force matrix
    forces = np.array([
        [np.cos(A), 0, 1, 1, 0, 0],
        [np.sin(A), 0, 0, 0, 1, 0],
        [-np.cos(A), np.cos(C), 0, 0, 0, 0],
        [-np.sin(A), -np.sin(C), 0, 0, 0, 0],
        [0, -np.cos(C), -1, 0, 0, 0],
        [0, np.sin(C), 0, 0, 0, 1]
    ])

    # load at point B
    loads = np.array([0, 0, 0, P, 0, 0])

    # solve for the forces
    solutions = np.linalg.solve(forces, loads)

    # save the forces
    total_forces = {"AB":0, "BC":0, "AC":0, "R_AX":0, "R_AY":0, "R_CY":0}

    total_forces["AB"] = solutions[0]
    total_forces["BC"] = solutions[1]
    total_forces["AC"] = solutions[2]
    total_forces["R_AX"] = solutions[3]
    total_forces["R_AY"] = solutions[4]
    total_forces["R_CY"] = solutions[5]

    # print the forces
    for force in total_forces:
        print("The force in", force, "=", format(total_forces[force], ".2f"), "lbs")

    return total_forces
```

## Find Coordinates of each point

```python

def calculate_triangle_coordinates(theta_A, theta_C, L_AC):

    # find angle B
    theta_B = 180 - theta_A - theta_C

    # change angles to radians
    A = np.radians(theta_A)
    B = np.radians(theta_B)
    C = np.radians(theta_C)

    # find the lengths
    length_AC = L_AC
    length_AB = (length_AC / np.sin(B)) * np.sin(C)
    length_BC = (length_AC / np.sin(B)) * np.sin(A)

    # print the lengths
    print("The length of AB =", format(length_AB, ".2f"), "ft")
    print("The length of BC =", format(length_BC, ".2f"), "ft")
    print("The length of AC =", format(length_AC, ".2f"), "ft")

    # find the coordinates
    joint_A = [0, 0]
    joint_B = [length_AB * np.cos(A), length_AB * np.sin(A)]
    joint_C = [length_AC, 0]

    # save the coordinates
    coordinates = {'A': joint_A, 'B': joint_B, 'C': joint_C}

    return coordinates
```

## Graph Truss  **Do Not Change Code**

```python
def graph_truss(coordinates, forces):
  # Each member connection with each starting, ending, and force value
  members = {("A", "B"): forces["AB"],
          ("B", "C"): forces["BC"],
          ("A", "C"): forces["AC"]}

  # Graph each member
  for start, end in members:
    force = members[start, end]
    if force > 0:
      color = "blue"  # sets the graphing color to be blue if in tension
      label = "Tension"
    else:
      color = "red" # sets graphing color to be red if in compression
      label = "Compression"
    start_x = coordinates[start][0]
    end_x = coordinates[end][0]
    start_y = coordinates[start][1]
    end_y = coordinates[end][1]
    x_values = [start_x, end_x]
    y_values = [start_y, end_y]
    plt.plot(x_values, y_values, color=color, label=label)


  # Graph each node
  for joint, location in coordinates.items():
    x = location[0]
    y = location[1]
    plt.scatter(x, y, color = "black")  # Black circle marker for joints
    plt.text(x*1.05, y*1.05, joint, fontsize=12) # added +.1 and +.2 for text to not overlap with joints


  # Add legend to lable tension and compression
  plt.legend()
  plt.margins(0.25)
  plt.grid(True)

  # Labels graph and x and y axes
  plt.xlabel("X-coordinate (ft) ")
  plt.ylabel("Y-coordinate (ft) ")
  plt.title("Truss Structure")
  plt.show()

graph_truss(calculate_triangle_coordinates(60,30,12), truss(60,30,-2000))
```
